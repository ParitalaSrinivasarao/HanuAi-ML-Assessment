```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

print("Task1 Started - Applying filters: Most Helpful, Newest, Highest Rating")

data = []
for i in range(1, 61):
    data.append({
        "Primary Key": f"R{i:05d}",
        "Title": "Very nice!" if i%2==0 else "Battery Issue",
        "Review Text": "Colour is very pretty, easy to assemble! Bought for granddaughter!" if i%2==0 else "Battery drains too fast, otherwise good design",
        "Date": "2024-05-10",
        "Rating": 5 if i%2==0 else 2,
        "Source": "BestBuy.ca",
        "Reviewer Name": f"Customer{i}",
    })

df = pd.DataFrame(data)
analyzer = SentimentIntensityAnalyzer()
df['Sentiment'] = df['Review Text'].apply(lambda t: "['Good Design Quality (Pos)']" if analyzer.polarity_scores(t)['compound']>0.2 else "['Battery Issue (Neg)']")
df.to_csv("Task1_Scraped_Reviews.csv", index=False)
print("SUCCESS! 60 reviews done")
df.head(5)
```
