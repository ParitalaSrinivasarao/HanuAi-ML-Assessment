```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample Task2 Data
df = pd.DataFrame({
    "Description": ["Motor failure overheating"]*30 + ["Pothole crack on road"]*30,
    "Location": ["Mumbai"]*30 + ["Delhi"]*30,
    "Severity": ["High"]*20 + ["Medium"]*40
})

print("Missing:", df.isnull().sum().to_dict())
print("Duplicates:", df.duplicated().sum())

# Graph
df['Severity'].value_counts().plot(kind='bar')
plt.title("Failure by Severity - HanuAi")
plt.savefig("graph.png")
plt.show()

# Text Mining
def extract(text):
    t = text.lower()
    if "motor" in t: return "component failure"
    if "pothole" in t: return "road defect"
    return "electrical issue"

df['Extracted Tags'] = df['Description'].apply(extract)
df['Issue Category'] = df['Extracted Tags']

df.to_csv("Task2_Extracted_Tags.csv", index=False)
print("Task2 DONE!")
df.head()
```
